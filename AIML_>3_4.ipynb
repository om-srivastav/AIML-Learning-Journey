```python
'''A college stores marks of 4 students in 3 subjects: Maths, Science, and English.

import numpy as np

marks = np.array([,
 ,
 ,
    [70, 76, 80]
])

Perform the following:

1. Display the marks of the third student.
2. Display the Science marks of the second student.
3. Add 5 bonus marks to every student's English marks.
4. Change the Maths marks of the fourth student to 75.
5. Display the final marks array.
'''
```

```python
import numpy as np

# Initializing the marks array
marks = np.array([
    [72, 85, 90],
    [65, 78, 82],
    [88, 91, 84],
    [70, 76, 80]
])

# 1. Display the marks of the third student (index 2)
print("1. Marks of the third student:")
print(marks[2])

# 2. Display the Science marks of the second student
# (Second student is row index 1, Science is column index 1)
print("\n2. Science marks of the second student:")
print(marks[1, 1])

# 3. Add 5 bonus marks to every student's English marks
# (English marks are in the last column, column index 2)
marks[:, 2] += 5

# 4. Change the Maths marks of the fourth student to 75
# (Fourth student is row index 3, Maths is column index 0)
marks[3, 0] = 75

# 5. Display the final marks array
print("\n5. Final marks array:")
print(marks)
```

```python
A company records monthly sales for 3 branches. The columns represent January, February, and March.

import numpy as np

sales = np.array([,
 ,
    [200, 220, 250]
])

Perform the following:
1. Display the complete sales data of Branch 2.
2. Display the sales of all branches in February.
3. Increase the March sales of every branch by 10%.
4. The January sales of Branch 1 were entered incorrectly. Change them from 120 to 135.
5. Display the updated sales array.
```

```python
import numpy as np

# Initializing the sales array as float to allow percentage calculations
sales = np.array([,  # Branch 1 (Jan, Feb, Mar),  # Branch 2 (Jan, Feb, Mar)
    [200, 220, 250]   # Branch 3 (Jan, Feb, Mar)
], dtype=float)

# --- Extra: Display data for ALL branches individually before changes ---
print("--- Initial Data for All Branches ---")
print("Branch 1 Sales Data:", sales[0])
print("Branch 2 Sales Data:", sales[1])
print("Branch 3 Sales Data:", sales[2])
print("-" * 38)

# 1. Display the complete sales data of Branch 2 (row index 1)
print("\n1. Complete sales data of Branch 2:")
print(sales[1])

# 2. Display the sales of all branches in February (column index 1)
print("\n2. February sales of all branches:")
print(sales[:, 1])

# 3. Increase the March sales of every branch by 10% (column index 2)
sales[:, 2] *= 1.10

# 4. Change January sales of Branch 1 from 120 to 135 (row index 0, column index 0)
sales[0, 0] = 135

```

```python
import numpy as np

employees = np.array([,
 ,
 ,
    [60000, 8]
])

Perform the following:
1. Display the information of the last employee using negative indexing.
2. Display the salary of the second employee.
3. Increase the salary of the third employee by 15%.
4. Add 1 year of experience to the first employee.
5. Display the final employee array.
```

```python
import numpy as np

# Initializing the employees array as float to allow percentage changes
employees = np.array([,
 ,
 ,
    [60000, 8]
], dtype=float)

# 1. Display the information of the last employee using negative indexing (-1)
print("1. Information of the last employee:")
print(employees[-1])

# 2. Display the salary of the second employee (row index 1, column index 0)
print("\n2. Salary of the second employee:")
print(employees[1, 0])

# 3. Increase the salary of the third employee by 15% (row index 2, column index 0)
employees[2, 0] *= 1.15

# 4. Add 1 year of experience to the first employee (row index 0, column index 1)
employees[0, 1] += 1

# 5. Display the final employee array
print("\n5. Final employee array:")
print(employees)
```

```python
A company has inventory information for 4 products across 3 warehouses.

Columns: Warehouse A, Warehouse B, Warehouse C

import numpy as np

inventory = np.array([,
 ,
 ,
    [55, 70, 90]
])

Perform the following:
1. Display the inventory of Product 3 across all warehouses.
2. Display the inventory of Warehouse B for all products.
3. Warehouse A receives 20 additional units of every product. Update the array.
4. The inventory of Product 4 in Warehouse C is incorrect. Change it to 100.
5. Display the final inventory.
```

```python
import numpy as np

# Initializing the inventory array
inventory = np.array([,  # Product 1,   # Product 2, # Product 3
    [55, 70, 90]     # Product 4
])

# 1. Display the inventory of Product 3 across all warehouses (row index 2)
print("1. Inventory of Product 3:")
print(inventory[2])

# 2. Display the inventory of Warehouse B for all products (column index 1)
print("\n2. Inventory of Warehouse B:")
print(inventory[:, 1])

# 3. Warehouse A receives 20 additional units of every product (column index 0)
inventory[:, 0] += 20

# 4. The inventory of Product 4 in Warehouse C is incorrect. Change it to 100.
# (Product 4 is row index 3, Warehouse C is column index 2)
inventory[3, 2] = 100

# 5. Display the final inventory
print("\n5. Final inventory array:")
print(inventory)
```
