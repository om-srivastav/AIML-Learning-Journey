#Catching Specific Errors

```python
try:
  n=int(input())
  print(100/n)
except Exception as e:#it will specify the error
  print(e)
```

```python
try:
  n=int(input())
  print(100/n)
  a=[3,4]
  print(a[4])#index error
except ZeroDivisionError:
  print("Division by 0 is not allowed")
except ValueError:
  print("Please enter numbers only")
except Exception as e:
  print("Some error has occured!")
```

```python
try:
  n=int(input())
  print(100/n)
  a=[3,4]
  print(a[4])
except Exception as e:
    print("some error")
except ZeroDivisionError:
    print("Division by 0")
except ValueError:
  print("pls enter number only")
```

```python
#else runs only when no exception occurs
try:
  num=int(input())
except ValueError:
  print("Invalid")
else:
  print("You Entered:",num)
```

```python
#finally block always runs
try:
  num=int(input())
except ValueError:
  print("Inavlid")
finally:
  print("Program run Successfully")
```
