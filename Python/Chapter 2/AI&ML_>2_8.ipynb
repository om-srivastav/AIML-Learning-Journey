```python
class Rectangle:
  def __init__(self,length,breadth):
    self.length=length
    self.breadth=breadth
  def area(self):
    return self.length + self.breadth
r= Rectangle(5,4)
print(r.area())
```

#Inheritance

```python
#parent class-->the class we r iheriting from{base class}
#child class-->it inherits properties, methods/features from the other class{derived class/sub class}
#Inheritance: it allows to inherit all the methods and properties from the parent class.
```

```python
class Person:
  def __init__(self,fname,lname):
    self.fname=fname
    self.lname=lname
  def show_name(self):
    print(self.fname,self.lname)
```

```python
p1= Person("Raj","Sharma")
p1.show_name()
```

```python
class Student(Person):
  pass
```

```python
s1=Student("Raj","Sharma")
```

```python
s1.show_name()
```
