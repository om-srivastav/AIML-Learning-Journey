```python
class Student:
  def __init__(self,name,age,city):
    self.name=name
    self.city=city
    self.age=age

  def display_profile(self):
    print(self.name,self.age,self.city)
```

```python
s1=Student("Om",19,"Deoria")
s2=Student("Rahul",22,"Mumbai")
s3=Student("Pankaj",55,"Kanpur")
```

```python
s1.display_profile()
```
