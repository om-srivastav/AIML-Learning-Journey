#P6--> Password Validator

```python
# a pass is valid if it contains
# minimum 8 chars
# atleast one uppercase
# atleast one lowercase
# atlease one digit

# and display
# strong or weak pass
```

```python
password = "omsri73hj"
upper=False
lower=False
digit=False
for c in password:
  if c.islower():
    lower=True
  elif c.isupper():
    upper=True
  elif c.isdigit():
    digit= True
n=len(password)
if n>=8 and upper==True and lower==True and digit==True:
  print("Strong pass")
else:
  print("weak")
```

#Cricket Scoreboard

```python
# take player name and runs
# display highest scorer
# and avg score
```

```python
scores={
    "Virat":90,
    "Rohit":95,
    "Dhoni":100,
    "Hardik":76,
    "Bumrah":20
}
type(scores)
```

```python
mx_score=0
for score in scores:
  if scores[score]>mx_score:
    mx_score = scores[score]
    mx_scorer = score
print(mx_score,mx_scorer)
```

```python
player=max(scores,key=scores.get)
print(player)
```
